# Quick start with the TDCRPy code

## Import the module

```python
import tdcrpy
```

## Enter the input parameters

```python
Rad="Co-60"                # radionuclide
pmf_1="1"                  # relative fraction of the radionulide
N = 1000                   # number of Monte Carlo trials
kB =1.0e-5                 # Birks constant in cm keV-1
V = 10                     # volume of scintillator in mL

lsCocktail = "Ultima Gold" # type of LS cocktail
micellCorr = True          # activate Micelle correction
frac_aq = 1/10             # aqueous fraction
solvant = "HCl"            # type of solvant
solvant_conc = 0.1         # concentration of solvant mol/L
t_W = 50                   # coincidence resolving time ns
t_EXT = 100                # extended dead time µs
quantum_effA = 0.25        # quantum efficiency of PMT A
quantum_effB = 0.25        # quantum efficiency of PMT B
quantum_effC = 0.25        # quantum efficiency of PMT C

# modification of advanced parameters
tdcrpy.TDCR_model_lib.modifyTau(int(t_W))
tdcrpy.TDCR_model_lib.modifyDeadTime(int(t_EXT))
tdcrpy.TDCR_model_lib.modifyEffQ(str(quantum_effA)+","+str(quantum_effB )+","+str(quantum_effC))
tdcrpy.TDCR_model_lib.modifySolvantType(solvant)
tdcrpy.TDCR_model_lib.modifySolvantConc(solvant_conc)
tdcrpy.TDCR_model_lib.modifyMicCorr(micellCorr)
tdcrpy.TDCR_model_lib.modifyLScocktail(lsCocktail, frac_aq, solvantType=solvant, solvantConc=solvant_conc)
```

## Detection efficiencies given the free parameter

```python
mode = "eff"               # ask for efficiency calculation
L = 5.1                    # free parameter in keV-1 (scintillation light yield) 
result1 = tdcrpy.TDCRPy.TDCRPy(L, Rad, pmf_1, N, kB, V, mode)
```

```python
print(f"efficiency S = {round(result1[0],4)} +/- {round(result1[1],4)}")
print(f"efficiency D = {round(result1[2],4)} +/- {round(result1[3],4)}")
print(f"efficiency T = {round(result1[4],4)} +/- {round(result1[5],4)}")
print(f"efficiency AB = {round(result1[6],4)} +/- {round(result1[7],4)}")
print(f"efficiency BC = {round(result1[8],4)} +/- {round(result1[9],4)}")
print(f"efficiency AC = {round(result1[10],4)} +/- {round(result1[11],4)}")
print(f"efficiency D (C/N system) = {round(result1[12],4)} +/- {round(result1[13],4)}")
```

## Full optical Monte Carlo transport

By default TDCRPy uses a semi-analytical detection model where photon yields
are treated as Poisson means and detection probabilities are computed
analytically.  Setting `opticalTransport=True` switches to a fully stochastic
optical model:

1. **Photon emission** — total photon count sampled from Poisson(*L* · *E_quenched*).
2. **Photon distribution** — photons distributed equally among PMTs
   ([1/3, 1/3, 1/3] for TDCR; [1/2, 1/2] for C/N).
3. **Photoelectron conversion** — per-PMT count drawn from Binomial(*n_photons*, *μ*).

Both models should agree within Monte Carlo uncertainty for symmetric
configurations.  The optical transport model adds variance and will become the
basis for geometry-aware photon transport in future versions.

## Full optical Monte Carlo transport

By default TDCRPy uses a **semi-analytical** detection model where the
expected number of photoelectrons per PMT is derived analytically from
the quenched energy and the free parameter *L*.

Setting `opticalTransport=True` activates a fully stochastic optical model:

1. **Photon emission** — total photon count drawn from Poisson(*L* · *E_q*).
2. **Photon distribution** — multinomial split: [1/3,1/3,1/3] for TDCR, [1/2,1/2] for C/N.
3. **Photoelectron conversion** — Binomial draw per PMT (quantum efficiency *μ*).

Both models agree within statistical uncertainty for symmetric configurations.

```python
import warnings; warnings.filterwarnings('ignore')

Rad_ot = 'Co-60'
L_ot   = 1.2
N_ot   = 2000   # smaller N for a quick comparison

# Semi-analytical detection model (default)
r_semi = tdcrpy.TDCRPy.TDCRPy(L_ot, Rad_ot, '1', N_ot, kB, V)

# Full optical MC transport
r_mc = tdcrpy.TDCRPy.TDCRPy(L_ot, Rad_ot, '1', N_ot, kB, V, opticalTransport=True)

print(f"{'':30s} {'Semi-analytical':>22}   {'Optical transport':>22}")
print('-' * 78)
for label, i in [('eff_S (single)', 0), ('eff_D (double)', 2), ('eff_T (triple)', 4)]:
    vs = f"{r_semi[i]:.4f} +/- {r_semi[i+1]:.4f}"
    vm = f"{r_mc[i]:.4f}  +/- {r_mc[i+1]:.4f}"
    print(f'  {label:<28} {vs:>22}   {vm:>22}')
```
