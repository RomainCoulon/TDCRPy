# Quick start with the TDCRPy code

## install a stable version of TDCRPy

```python
pip install TDCRPy==2.17.2
```

## Import the module

```python
import tdcrpy
```

## Enter the input parameters

```python
Rad="Co-60"                # radionuclide
pmf_1="1"                  # relative fraction of the radionulide
N = 1000                   # number of Monte Carlo trials
kB =1.0e-5                 # Birks constant in cm keV-1
V = 10                     # volume of scintillator in mL
lsCocktail = "Ultima Gold" # type of LS cocktail
frac_aq = 1/10             # aqueous fraction
solvant = "HCl"            # type of solvant
solvant_conc = 0.1         # concentration of solvant mol/L
t_W = 50                   # coincidence resolving time ns
t_EXT = 100                # extended dead time µs
quantum_effA = 0.25        # quantum efficiency of PMT A
quantum_effB = 0.25        # quantum efficiency of PMT B
quantum_effC = 0.25        # quantum efficiency of PMT C

# modification of advanced parameters
tdcrpy.TDCR_model_lib.modifyTau(int(t_W))
tdcrpy.TDCR_model_lib.modifyDeadTime(int(t_EXT))
tdcrpy.TDCR_model_lib.modifyEffQ(str(quantum_effA)+","+str(quantum_effB )+","+str(quantum_effC))
tdcrpy.TDCR_model_lib.modifySolvantType(solvant)
tdcrpy.TDCR_model_lib.modifySolvantConc(solvant_conc)
tdcrpy.TDCR_model_lib.modifyLScocktail(lsCocktail, frac_aq, solvantType=solvant, solvantConc=solvant_conc)
```

## Detection efficiencies given the free parameter

```python
mode = "eff"               # ask for efficiency calculation
L = 5.1                    # free parameter in keV-1 (scintillation light yield) 
result1 = tdcrpy.TDCRPy.TDCRPy(L, Rad, pmf_1, N, kB, V, mode)
```

```python
print(f"efficiency S = {round(result1[0],4)} +/- {round(result1[1],4)}")
print(f"efficiency D = {round(result1[2],4)} +/- {round(result1[3],4)}")
print(f"efficiency T = {round(result1[4],4)} +/- {round(result1[5],4)}")
print(f"efficiency AB = {round(result1[6],4)} +/- {round(result1[7],4)}")
print(f"efficiency BC = {round(result1[8],4)} +/- {round(result1[9],4)}")
print(f"efficiency AC = {round(result1[10],4)} +/- {round(result1[11],4)}")
print(f"efficiency D (C/N system) = {round(result1[12],4)} +/- {round(result1[13],4)}")
```

## Detection efficiencies given the measured TDCR parameter

### Case of global TDCR parameter

```python
TD = 0.977667386529166     # TDCR parameter
minL = 0.01                # minimum light yield (keV-1)
maxL = 20                  # maximum light yield (keV-1)
```

```python
result2 = tdcrpy.TDCRPy.eff(TD, Rad, pmf_1, kB, V, N, Lbounds=(minL, maxL))
```

```python
print(f"global free parameter = {round(result2[0],4)} keV-1")
print(f"global free parameter (PMT A) = {round(result2[1][0],4)} keV-1")
print(f"global free parameter (PMT B) = {round(result2[1][1],4)} keV-1")
print(f"global free parameter (PMT C) = {round(result2[1][2],4)} keV-1")
print(f"efficiency S = {round(result2[2],4)} +/- {round(result2[3],4)}")
print(f"efficiency D = {round(result2[4],4)} +/- {round(result2[5],4)}")
print(f"efficiency T = {round(result2[6],4)} +/- {round(result2[7],4)}")
print(f"efficiency AB = {round(result2[6],4)} +/- {round(result2[7],4)}")
print(f"efficiency BC = {round(result2[8],4)} +/- {round(result2[9],4)}")
print(f"efficiency AC = {round(result2[10],4)} +/- {round(result2[11],4)}")
print(f"efficiency D (C/N system) = {round(result2[14],4)} +/- {round(result2[15],4)}")
```

### Case of specific TDCR parameters

```python
TD = 0.977667386529166     # TDCR parameter
TAB = 0.992232838598821    # TDCR parameter T/AB
TBC = 0.992343419459002    # TDCR parameter T/BC
TAC = 0.99275350064608     # TDCR parameter T/AC
minL = 0.01                # minimum light yield (keV-1)
maxL = 20                  # maximum light yield (keV-1)
```

```python
result3 = tdcrpy.TDCRPy.eff([TD, TAB, TBC, TAC], Rad, pmf_1, kB, V, N, Lbounds=(minL, maxL))
```

```python
print(f"Global free parameter = {round(result3[0],4)} keV-1")
print(f"free parameter of PMT A = {round(result3[1][0],4)} keV-1")
print(f"free parameter of PMT B = {round(result3[1][1],4)} keV-1")
print(f"free parameter of PMT C = {round(result3[1][2],4)} keV-1")
print(f"efficiency S = {round(result3[2],4)} +/- {round(result3[3],4)}")
print(f"efficiency D = {round(result3[4],4)} +/- {round(result3[5],4)}")
print(f"efficiency T = {round(result3[6],4)} +/- {round(result3[7],4)}")
print(f"efficiency AB = {round(result3[8],4)} +/- {round(result3[9],4)}")
print(f"efficiency BC = {round(result3[10],4)} +/- {round(result3[11],4)}")
print(f"efficiency AC = {round(result3[12],4)} +/- {round(result3[13],4)}")
print(f"efficiency D (C/N system) = {round(result3[14],4)} +/- {round(result3[15],4)}")
```
