```python
import tdcrpy as td
import numpy as np
from tqdm import tqdm
import matplotlib.pyplot as plt
```

```python
Rad="Sr-90, Y-90"           # radionuclides
pmf_1="0.5, 0.5"                # relatives fractions of the radionulides
kB =1.0e-5                  # Birks constant in cm keV-1
V = 10                      # volume of scintillator in mL
```

```python
TD = 0.9838                # TDCR parameter
minL = 0.1                 # minimum light yield (keV-1)
maxL = 100                 # maximum light yield (keV-1)
N = 10000                  # number of Monte-Carlo trials
result2 = td.TDCRPy.eff(TD, Rad, pmf_1, kB, V, N, Lbounds=(minL, maxL))
```

```python
print(f"global free parameter = {round(result2[0],4)} keV-1")
print(f"global free parameter (PMT A) = {round(result2[1][0],4)} keV-1")
print(f"global free parameter (PMT B) = {round(result2[1][1],4)} keV-1")
print(f"global free parameter (PMT C) = {round(result2[1][2],4)} keV-1")
print(f"efficiency S = {2*round(result2[2],4)} +/- {2*round(result2[3],4)}")
print(f"efficiency D = {2*round(result2[4],4)} +/- {2*round(result2[5],4)}")
print(f"efficiency T = {2*round(result2[6],4)} +/- {2*round(result2[7],4)}")
print(f"efficiency AB = {2*round(result2[6],4)} +/- {2*round(result2[7],4)}")
print(f"efficiency BC = {2*round(result2[8],4)} +/- {2*round(result2[9],4)}")
print(f"efficiency AC = {2*round(result2[10],4)} +/- {2*round(result2[11],4)}")
```
